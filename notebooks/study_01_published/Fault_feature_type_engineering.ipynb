```python
from pathlib import Path
import pandas as pd

extract_root = Path("/content/step3_step3_campaign_v1_extract")
merged_path = extract_root / "step3_merged_production_only.csv"

df = pd.read_csv(merged_path)

print("Total rows:", len(df))

# One canonical condition key per row
df["condition_key"] = df["fault_type"].astype(str) + "_s" + df["fault_severity"].astype(int).astype(str)

expected_conditions = {
    "none_s0",
    "traffic_overload_s1",
    "traffic_overload_s2",
    "traffic_overload_s3",
    "radio_degradation_s1",
    "radio_degradation_s2",
    "radio_degradation_s3",
}

# Collapse to one record per (seed, run, condition)
block_condition = (
    df[["seed", "run_id", "condition_key"]]
    .drop_duplicates()
    .sort_values(["seed", "run_id", "condition_key"])
)

# Count distinct conditions per block
block_summary = (
    block_condition.groupby(["seed", "run_id"])["condition_key"]
    .agg(lambda x: sorted(set(x)))
    .reset_index()
)

block_summary["condition_count"] = block_summary["condition_key"].apply(len)
block_summary["is_complete_7of7"] = block_summary["condition_count"] == 7
block_summary["missing_conditions"] = block_summary["condition_key"].apply(
    lambda x: sorted(expected_conditions - set(x))
)

print("\n=== BLOCK COMPLETENESS SUMMARY ===")
print("Retained blocks found:", len(block_summary))
print("Complete 7/7 blocks  :", int(block_summary["is_complete_7of7"].sum()))
print("Incomplete blocks    :", int((~block_summary["is_complete_7of7"]).sum()))

incomplete = block_summary[~block_summary["is_complete_7of7"]].copy()

if len(incomplete) > 0:
    print("\n=== INCOMPLETE BLOCKS FOUND ===")
    for _, row in incomplete.iterrows():
        print(f"(seed={row['seed']}, run={row['run_id']}): missing {row['missing_conditions']}")
else:
    print("\nNo incomplete retained blocks found.")

# Strict checks
assert len(block_summary) == 33, f"Expected 33 retained blocks, found {len(block_summary)}"
assert int(block_summary["is_complete_7of7"].sum()) == 33, \
    "Not all retained blocks are methodologically complete"
assert len(incomplete) == 0, \
    "One or more retained blocks are incomplete"

# Also verify exactly 2 rows per (seed, run, condition)
rows_per_block_condition = (
    df.groupby(["seed", "run_id", "condition_key"])
      .size()
      .reset_index(name="rows")
      .sort_values(["seed", "run_id", "condition_key"])
)

bad_row_units = rows_per_block_condition[rows_per_block_condition["rows"] != 2].copy()

print("\n=== PER (seed, run, condition) ROW CHECK ===")
print("Units checked:", len(rows_per_block_condition))
print("Bad units    :", len(bad_row_units))

if len(bad_row_units) > 0:
    print("\nBad units:")
    print(bad_row_units.to_string(index=False))

assert len(rows_per_block_condition) == 231, \
    f"Expected 231 retained block-condition units, found {len(rows_per_block_condition)}"
assert len(bad_row_units) == 0, \
    "Some retained (seed, run, condition) units do not have exactly 2 rows"

print("\n[OK] Matched-block completeness consistency passed.")
```

```python
from pathlib import Path
import pandas as pd
from sklearn.model_selection import GroupShuffleSplit

extract_root = Path("/content/step3_step3_campaign_v1_extract")
ml_path = extract_root / "step3_ml_ready_fault_type_baseline_v1.csv"

df = pd.read_csv(ml_path)

# Features / target / grouping
feature_cols = [
    "tx_packets",
    "rx_packets",
    "tx_bytes",
    "rx_bytes",
    "throughput_mbps",
    "mean_delay_ms",
    "mean_jitter_ms",
    "time_last_rx_s",
    "flow_duration_s",
]

target_col = "fault_type"
group_cols = ["seed", "run_id"]

# Build group key
df["group_key"] = df["seed"].astype(str) + "_run" + df["run_id"].astype(str)

X = df[feature_cols].copy()
y = df[target_col].copy()
groups = df["group_key"].copy()

# Group-safe split
gss = GroupShuffleSplit(n_splits=1, test_size=0.20, random_state=42)
train_idx, test_idx = next(gss.split(X, y, groups=groups))

train_df = df.iloc[train_idx].copy()
test_df  = df.iloc[test_idx].copy()

# Save split files
train_path = extract_root / "step3_ml_train_fault_type_baseline_v1.csv"
test_path  = extract_root / "step3_ml_test_fault_type_baseline_v1.csv"

train_df.to_csv(train_path, index=False)
test_df.to_csv(test_path, index=False)

print("=== GROUP-SAFE SPLIT COMPLETE ===")
print("Train file:", train_path)
print("Test file :", test_path)

print("\n=== SHAPES ===")
print("Train rows:", train_df.shape[0])
print("Test rows :", test_df.shape[0])

print("\n=== UNIQUE BLOCK COUNTS ===")
print("Train blocks:", train_df[group_cols].drop_duplicates().shape[0])
print("Test blocks :", test_df[group_cols].drop_duplicates().shape[0])

print("\n=== GROUP OVERLAP CHECK ===")
train_groups = set(train_df["group_key"].unique())
test_groups = set(test_df["group_key"].unique())
overlap = train_groups.intersection(test_groups)
print("Overlap groups:", len(overlap))
print("Overlap set:", overlap)

print("\n=== TARGET COUNTS: TRAIN ===")
print(train_df[target_col].value_counts())

print("\n=== TARGET COUNTS: TEST ===")
print(test_df[target_col].value_counts())
```

```python
import numpy as np
import pandas as pd

from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (
    accuracy_score,
    f1_score,
    classification_report,
    confusion_matrix
)

# ============================================================
# LOCKED FEATURE SET
# ============================================================

refined_features = [
    "tx_packets",
    "rx_packets",
    "tx_bytes",
    "rx_bytes",
    "throughput_mbps",
    "mean_delay_ms",
    "mean_jitter_ms",
    "time_last_rx_s",
    "flow_duration_s",
    "packet_loss_ratio",
    "jitter_delay_ratio",
    "loss_jitter_interaction",
]

X_train_mlp = X_train[refined_features].copy()
X_test_mlp = X_test[refined_features].copy()

# ============================================================
# LABEL ENCODING FOR MLP ONLY
# ============================================================

label_encoder = LabelEncoder()
y_train_enc = label_encoder.fit_transform(y_train)
y_test_enc = label_encoder.transform(y_test)

print("=== Label Mapping ===")
for cls, idx in zip(label_encoder.classes_, range(len(label_encoder.classes_))):
    print(f"{cls} -> {idx}")
print()

# ============================================================
# CONTROLLED MINIMAL MLP BASELINE
# ============================================================

mlp_pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("mlp", MLPClassifier(
        hidden_layer_sizes=(16,),
        activation="relu",
        solver="adam",
        alpha=1e-3,
        batch_size=16,
        learning_rate_init=1e-3,
        max_iter=500,
        early_stopping=True,
        validation_fraction=0.15,
        n_iter_no_change=20,
        random_state=42
    ))
])

mlp_pipeline.fit(X_train_mlp, y_train_enc)

y_pred_enc = mlp_pipeline.predict(X_test_mlp)
y_proba = mlp_pipeline.predict_proba(X_test_mlp)

# back to original class names
y_pred = label_encoder.inverse_transform(y_pred_enc)

# ============================================================
# METRICS
# ============================================================

acc = accuracy_score(y_test, y_pred)
macro_f1 = f1_score(y_test, y_pred, average="macro")

label_order = ["none", "radio_degradation", "traffic_overload"]

cm = confusion_matrix(y_test, y_pred, labels=label_order)
report = classification_report(
    y_test,
    y_pred,
    labels=label_order,
    digits=6,
    zero_division=0
)

print("=== Minimal MLP Baseline ===")
print(f"Accuracy: {acc:.6f}")
print(f"Macro F1: {macro_f1:.6f}")
print()

print("=== Classification Report ===")
print(report)

print("=== Confusion Matrix ===")
cm_df = pd.DataFrame(
    cm,
    index=[f"true_{c}" for c in label_order],
    columns=[f"pred_{c}" for c in label_order]
)
print(cm_df)
print()

# ============================================================
# TRAINING DIAGNOSTICS
# ============================================================

mlp_model = mlp_pipeline.named_steps["mlp"]

print("=== Training Diagnostics ===")
print("n_iter_:", mlp_model.n_iter_)
print("loss_:", mlp_model.loss_)
print("best_validation_score_:", getattr(mlp_model, "best_validation_score_", None))
print()

# ============================================================
# PROBABILITY / MARGIN ANALYSIS
# Need class-name columns, not encoded integers
# ============================================================

class_names = label_encoder.inverse_transform(np.arange(len(label_encoder.classes_)))
proba_df = pd.DataFrame(y_proba, columns=class_names, index=X_test_mlp.index)

sorted_probs = np.sort(y_proba, axis=1)
top1 = sorted_probs[:, -1]
top2 = sorted_probs[:, -2]
margin = top1 - top2

confidence_df = pd.DataFrame({
    "y_true": y_test.values,
    "y_pred": y_pred,
    "top1_confidence": top1,
    "top2_confidence": top2,
    "margin": margin,
    "correct": (y_test.values == y_pred).astype(int)
}, index=X_test_mlp.index)

confidence_df = pd.concat([confidence_df, proba_df.add_prefix("proba_")], axis=1)

print("=== Confidence Summary by TRUE Class ===")
summary_true = confidence_df.groupby("y_true").agg(
    n=("margin", "count"),
    mean_top1_confidence=("top1_confidence", "mean"),
    median_top1_confidence=("top1_confidence", "median"),
    mean_margin=("margin", "mean"),
    median_margin=("margin", "median"),
    accuracy_within_class=("correct", "mean")
).sort_index()
print(summary_true.to_string())
print()
```
